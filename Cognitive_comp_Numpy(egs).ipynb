Q1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:
a. Add 2°C to every reading using vectorization.
b. Convert all temperatures from Celsius to Fahrenheit using: F = (9/5)C+32
c. Identify all readings greater than 32°C using Boolean indexing.
d. Count how many readings exceed 32°C.
e. Explain how vectorization and Boolean indexing can make data processing more efficient
than explicitly iterating through every value.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temp_2 = temperature+2
print(temp_2)

temp_fahrenheit = ((9/5)*temp_2) + 32
print(temp_fahrenheit)

temp_3 = temperature[temperature > 32]
print(temp_3)

print(temp_3.size)
#or
count_temp_3 = np.sum(temperature > 32)
print(count_temp_3)

'''Boolean indexing selects all elements satisfying a condition at once,
without using a loop to check each value individually.'''
'''Vectorization performs operations on the entire array at once,
making processing faster and more efficient than explicit iteration.'''
```

Q2. The following matrix represents the number of daily steps recorded by four users
over three days:
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])
Perform the following:
a. Calculate the total steps recorded.
b. Calculate the mean number of steps.
c. Find the maximum and minimum values.
d. Calculate the total steps for each day using axis=0.
e. Calculate the total steps for each user using axis=1.
f. Find the user/day position containing the maximum number of steps using argmax().

```python
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

daily_total = np.sum(steps, axis=0)
print("Total steps for each day:", daily_total)

user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)

position = np.unravel_index(np.argmax(steps), steps.shape)

print("Maximum value:", steps[position])
print("User index:", position[0])
print("Day index:", position[1])
```

Q3. Write a Python program using NumPy to perform the following operations:
a. Create the following NumPy array:
original = np.array([1, 2, 3, 4, 5, 6])
b. Create a slice containing elements from index 1 to 4 and store it in a variable named subset.
c. Modify the first element of subset to 999 and display both original and subset.
d. Create another slice from original, but this time use .copy(). Modify its first element to 500
and display original and the copied array.
e. Create a NumPy array containing the numbers from 1 to 12 using np.arange() and reshape it
into a 3 × 4 matrix.
f. From the 3 × 4 matrix, extract the following using NumPy indexing and slicing:
• First row
• Last row
• Second column
• Elements from rows 1–2 and columns 2–3
g. Create a flattened version of the 3 × 4 matrix using both flatten() and ravel().
h. Modify an element in the array returned by ravel() and observe whether the original array is
affected.
i. Modify an element in the array returned by flatten() and observe whether the original array is
affected.
j. Display the shape, ndim, size, and dtype of the original 3 × 4 matrix.

```python
import numpy as np

# (a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# (b) Create a slice from index 1 to 4
subset = original[1:5]
print("Subset:", subset)

# (c) Modify the first element of the subset
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

# (d) Create a separate copy and modify it
copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original after modifying copy:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

# First row
print("First row:", matrix[0, :])

# Last row
print("Last row:", matrix[-1, :])

# Second column
print("Second column:", matrix[:, 1])

# Rows 1–2 and columns 2–3 (using one-based numbering)
print("Submatrix:\n", matrix[0:2, 1:3])

flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("Flatten:", flat_array)
print("Ravel:", ravel_array)

ravel_array[0] = 100

print("Ravel array:", ravel_array)
print("Original matrix:\n", matrix)

flat_array[1] = 200

print("Flatten array:", flat_array)
print("Original matrix:\n", matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based
on three factors:
• Sleep hours
• Activity level
• Stress level
Consider:
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])
Here, each row represents one user and each column represents a feature.
Perform the following:
a. b. c. d. f. Display the shape and dimensions of X.
Calculate: X.T and explain what the transpose represents.
Calculate the matrix product: X.T @ X
Calculate the matrix inverse using: np.linalg.inv() where appropriate.
e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋!𝑋)"#𝑋!𝑦 using NumPy.
Explain what the resulting coefficients represent in the context of the three user
features.
g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to
obtain a predicted assistance score.

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T

print("Transpose of X:\n", XT)

XTX = X.T @ X

print("X transpose multiplied by X:\n", XTX)

XTX_inv = np.linalg.inv(XTX)

print("Inverse of XTX:\n", XTX_inv)

beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("OLS coefficients:", beta)

'''The coefficients correspond to the three features in the order they appear in X.
Feature	Coefficient (approx.)	Interpretation
Sleep hours	-0.1592	A one-unit increase in sleep hours corresponds to a decrease of about 0.159 in the predicted score, holding the other features constant.
Activity level	0.1408	A one-unit increase in activity level corresponds to an increase of about 0.141 in the predicted score, holding other features constant.
Stress level	10.0630	A one-unit increase in stress corresponds to an increase of about 10.063 in the predicted score, holding other features constant.
These are mathematical interpretations of the fitted coefficients, not evidence that these factors cause the score to change. The model is fitted to only four observations.
Also, the specified OLS equation has no intercept term, because the assignment provides the formula without adding a column of ones to X.'''

new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
